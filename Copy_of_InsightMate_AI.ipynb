```python
import pandas as pd
import matplotlib.pyplot as plt

print("InsightMate AI data-analysis tools are ready!")
```

```python
data = {
    "Respondent": ["P01", "P02", "P03", "P04", "P05"],
    "Age": [22, 25, 21, 30, 27],
    "Hours_Studied": [2, 4, 3, 5, 4],
    "Test_Score": [60, 75, 68, 90, 82]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("INSIGHTMATE AI: DATA SUMMARY")
print("-----------------------------")

print("Average age:", df["Age"].mean())
print("Average study hours:", df["Hours_Studied"].mean())
print("Average test score:", df["Test_Score"].mean())
print("Highest test score:", df["Test_Score"].max())
print("Lowest test score:", df["Test_Score"].min())
```

```python
print("INSIGHTMATE AI: WHAT THE DATA TELLS US")
print("--------------------------------------")

average_score = df["Test_Score"].mean()
average_hours = df["Hours_Studied"].mean()

print(f"The average test score is {average_score:.1f}.")
print(f"Students studied for an average of {average_hours:.1f} hours.")
print("The highest test score was 90.")
print("The lowest test score was 60.")

if average_score >= 70:
    print("Overall, the average test score is 70 or above.")
else:
    print("Overall, the average test score is below 70.")
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df["Respondent"], df["Test_Score"])

plt.title("Test Scores by Respondent")
plt.xlabel("Respondent")
plt.ylabel("Test Score")

plt.ylim(0, 100)
plt.show()
```

```python
print("INSIGHTMATE AI: DATA QUALITY CHECK")
print("----------------------------------")

print("Missing values in each column:")
print(df.isnull().sum())

if df.isnull().sum().sum() == 0:
    print("Good news! No missing values were found.")
else:
    print("Attention: Some values are missing. Please review the data.")
```

```python
test_data = df.copy()

test_data.loc[0, "Test_Score"] = None

print("Missing test scores:", test_data["Test_Score"].isnull().sum())

if test_data["Test_Score"].isnull().sum() > 0:
    print("Warning: Missing test scores detected.")
else:
    print("No missing test scores found.")
```
